```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 基础配置（和你原有代码一致）
ROOT_DIR = '/data2/code/Project/GC/downstream/GRASP/PLOTS/2.km/final/临床&Risk信息'
centers = ['SXCH-Train', 'SXCH-Val', 'External', 'YYH', 'SYSUCC', 'TCGA_STAD']

# 设置中文字体（解决中文乱码，不需要可以删除）
# plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
# plt.rcParams["axes.unicode_minus"] = False  # 负号正常显示

# 读取所有中心的数据
all_data = []
for center in centers:
    file_path = os.path.join(ROOT_DIR, f'{center}_Info.csv')
    df = pd.read_csv(file_path, dtype={'case_id': str})
    df['center'] = center  # 增加一列标记中心
    all_data.append(df[['risk', 'center']])  # 只保留需要的列

# 合并成一个大表
df_all = pd.concat(all_data, ignore_index=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 基础配置

# 字体配置（无警告）
plt.rcParams["font.family"] = ["DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False

# ===================== 绘图 1：分中心子图 =====================
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(14, 12))
axes = axes.flatten()

for i, center in enumerate(centers):
    print(f'当前中心: {center}')
    ax = axes[i]
    data = df_all[df_all['center'] == center]['risk']
    
    sns.histplot(data=data, kde=True, bins=30, ax=ax, color='#2E86AB', alpha=0.7)
    ax.set_title(f'{center} - Risk Distribution', fontsize=12, fontweight='bold')
    ax.set_xlabel('Risk Score', fontsize=10)
    ax.set_ylabel('Frequency', fontsize=10)
    ax.grid(alpha=0.3)

# axes[-1].set_visible(False)
plt.tight_layout()
plt.savefig(os.path.join(ROOT_DIR, 'Risk_Distribution_Subplots.png'), dpi=300, bbox_inches='tight')
plt.close()

# ===================== 绘图 2：叠加对比图（修复图例警告）=====================
plt.figure(figsize=(12, 6))

# 核心修复：指定 hue 顺序 + 确保图例正常生成
kde_plot = sns.kdeplot(
    data=df_all, 
    x='risk', 
    hue='center', 
    hue_order=centers,  # 强制指定中心顺序
    linewidth=2.5, 
    fill=True, 
    alpha=0.2, 
    palette='tab10'
)

plt.title('Risk Score Distribution Across Centers', fontsize=14, fontweight='bold')
plt.xlabel('Risk Score', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(alpha=0.3)

# 修复：自动获取已有图例，不手动生成空图例
plt.tight_layout()

plt.savefig(os.path.join(ROOT_DIR, 'Risk_Distribution_Comparison.png'), dpi=300, bbox_inches='tight')
plt.close()

print("✅ 绘图完成！无任何警告！文件已保存到：", ROOT_DIR)
```
